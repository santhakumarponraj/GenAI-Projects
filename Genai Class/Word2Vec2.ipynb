```python
reviews = [
    "The movie was fantastic and I loved it",
    "What a great film, amazing acting",
    "I hated the movie, it was boring",
    "The plot was dull and the acting was bad",
    "An excellent and thrilling experience",
    "Terrible movie, waste of time",
    "I really enjoyed the story and characters",
    "Worst film ever, not recommended"
]
```

```python
import re

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()

processed_reviews = [preprocess(r) for r in reviews]

print(processed_reviews)
```

```python
# Train Word2Vec Model
from gensim.models import Word2Vec

model = Word2Vec(
    sentences=processed_reviews,
    vector_size=100,
    window=5,
    min_count=1
)
```

```python
print(model.wv['movie'])
```

```python
print(model.wv.most_similar('movie'))
```

```python
print(model.wv.most_similar('great'))
```

```python

print(model.wv.most_similar('terrible'))
```

```python
print(model.wv.similarity('great', 'amazing'))
print(model.wv.similarity('great', 'bad'))
```

```python
print(model.wv.most_similar(positive=['great','bad'], negative=['terrible']))
```

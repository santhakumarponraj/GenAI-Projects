# Simple Generative AI Python Program

```python
import google.generativeai as genai

# Add your Gemini API key
genai.configure(api_key= "key")

for model in genai.list_models():
    print(model.name)
```

# Short motivational story about a student learning Artificial Intelligence

```python
# load Gemini model
model = genai.GenerativeModel("models/gemini-2.5-flash")

# give prompt
response = model.generate_content(
    "Write a short motivational story about a student learning Artificial Intelligence."
)

# print AI generated text
print(response.text)
```
